```python
import pandas as pd
from defenders.tools.code.src.code_defender import CodeDefender
from sklearn.metrics import classification_report
```

```python
df_benign=pd.read_csv('./../data/benign_data.csv')
df_malicious=pd.read_csv('./../data/malicious_data.csv')
```

```python
df_all=pd.concat([df_benign,df_malicious],ignore_index=True)
```

```python
code_defender = CodeDefender(deep_check=False)
```

```python
labels = []
true_labels = []
for index, row in df_all.iterrows():
    code = row['code']
    label = row['label']
    result,message = code_defender.is_safe(code)
    labels.append(1-result)
    true_labels.append(label)
```

```python
print(classification_report(true_labels, labels))
```
